

Basic Example of pyCATCH usage
===============================

This example notebook details the basic usage of pyCATCH.
It documents the process of coronal hole extraction from downloading the data, performing the analysis and saving the results.



```python
#THIS IS NEEDED FOR RUNNING PYCATCH IN A JUPYTER NOTEBOOK
%matplotlib widget 
# import the pycatch class
from pycatch.pycatch import pycatch
```

```python
# initialize pycatch 
ch = pycatch()
```

```python
# download EUV data 
# default is AIA/SDO 193 data
# default channel is JSOC; needs  JSOC registered email
ch.download('2013-05-29T18:00', email='example@pycatch.com')
```

```python
# load EUV data 
# filepath of the downloaded file is stored in ch.map_file, it can also be given manually
ch.load()
```

```python
# download magnetogram 
# default is HMI/SDO 720s data
# EUV map must be loaded for the magnetogram download to work, it downloads the magnetogram date closest to the EUV map date
ch.download_magnetogram(email='example@pycatch.com')
```

```python
# load magnetic field data 
# filepath of the downloaded file is stored in ch.magnetogram_file, it can also be given manually
ch.load(mag=True)
```

```python
# quicklook at loaded images (use mag=True for magnetogram)
ch.show()
```

```python
# calibrate data
# EUV: Default behaviour: register, normalize,  correct for instrument degradation, annulus limb correction, set off limb pixel to nan 
ch.calibration()
```

```python
# magnetogram: Default behaviour: align with EUV map, set off limb pixel to nan 
ch.calibration_mag()
```

```python
# reduce resolution
# Default: bin to 1024x1024
# rebins both the EUV map and the magnetogram if both are loaded
ch.rebin()
```

```python
# cutout a sub region of the sun
# Default: cuts the map from (-1100,-1100) arsec to (1100,1100) arcsec 
ch.cutout()
```

```python
# select points within coronal holes that should be extracted
# hint displays possible coronal holes
ch.select(hint=True)
```

```python
# find threshold for coronal hole extractions (options can be found in the documentation)
ch.suggest_threshold() #
```

```python
# extract the coronal holes
ch.extract_ch()
```

```python
# view coronal hole extraction
# EUV map
ch.plot_map()
# magnetogram
ch.plot_map(mag=True)
```

```python
# calculate coronal hole properties
# morphological properties
ch.calculate_properties()
```

```python
# magnetic properties
ch.calculate_properties(mag=True)
```

```python
# save catch results
# save properties in text files
ch.print_properties()
```

```python
# save pycatch object // save your progress
ch.save()
```

```python
# save plots
# EUV map
ch.plot_map(save=True)
# magnetogram
ch.plot_map(mag=True,save=True)
```

```python
# save binary coronal hole map to fits file
ch.save_map()
```
